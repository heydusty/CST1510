## 3. `from module import name`

Brings in one specific name directly — no dot needed afterwards. Convenient, but a
reader can no longer tell at a glance which module it came from.

````python
from math import sqrt

print(sqrt(16))   # not math.sqrt(16)
````

## 5. Making your own file both a module and a script

Every mini-project you write is a `.py` file — which makes it a module too. If
someone else `import`s it, Python runs every top-level line immediately,
including any test `print()`s you left in. The `__name__ == "__main__"` guard
lets you say "only run this part when this file is launched directly."

````python
def area_of_circle(radius):
    return math.pi * radius ** 2

if __name__ == "__main__":
    print(area_of_circle(5))
````

```
python this_file.py         ──▶  __name__ is "__main__"   ──▶  the block runs
import this_file             ──▶  __name__ is "this_file"    ──▶  it does NOT run
```

The function definition above the guard runs either way — only the code
*inside* the `if` is skipped when the file is imported instead of run.

## Common mistakes

| Mistake | What you see | Fix |
|---|---|---|
| misspelling the module name — `import maths` | `ModuleNotFoundError: No module named 'maths'` | it is `math`, no `s` |
| calling `sqrt(16)` after `import math` (not `from`) | `NameError` | either `math.sqrt(16)` or `from math import sqrt` |
| a `from import` overwriting your own variable of the same name | wrong value, no error | check your variable names against what you are importing |
| forgetting the module needs installing | `ModuleNotFoundError` for a real package name | `pip install <package>` first, then `import` |
| test code at the top level, no `__name__` guard | it runs every time someone `import`s your file, not just when you run it | put it inside `if __name__ == "__main__":` |
